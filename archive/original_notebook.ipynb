```python
import os
import csv
from email.parser import Parser

# Define the root directory of the Enron dataset
root_dir = "maildir"

# Define the output CSV file
output_csv = 'enron_data.csv'

# Initialize the CSV writer
with open(output_csv, 'w', newline='', encoding='utf-8') as csvfile:
    writer = csv.writer(csvfile)
    writer.writerow([0,1,2,3,4,5,6])  # Header row

    # Iterate through each person's folder
    for person in os.listdir(root_dir):
        person_dir = os.path.join(root_dir, person)
        if os.path.isdir(person_dir):
            documents_dir = os.path.join(person_dir, 'inbox')
            if os.path.isdir(documents_dir):
                # Iterate through each document in the all-documents folder
                for document in os.listdir(documents_dir):
                    document_path = os.path.join(documents_dir, document)
                    if document_path.endswith('.'):
                        try:
                             with open(document_path, 'r', encoding='utf-8') as file:
                                email_content = file.read()
                                parser = Parser()
                                email_message = parser.parsestr(email_content)

                                # Extract fields
                                message_id = email_message['Message-ID']
                                date = email_message['Date']
                                from_addr = email_message['From']
                                to_addr = email_message['To']
                                subject = email_message['Subject']

                                # Extract the body of the email
                                body = ''
                                if email_message.is_multipart():
                                    for part in email_message.walk():
                                        if part.get_content_type() == 'text/plain':
                                            body = part.get_payload(decode=True).decode('utf-8')
                                            break
                                else:
                                    body = email_message.get_payload(decode=True).decode('utf-8') if email_message.get_payload(decode=True) else ''
                                writer.writerow([person, message_id, date, from_addr, to_addr, subject, body])
                        except Exception as e:
                            print(f"Error reading {document_path}: {e}")
```
